```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid

print("🔭 Gravitational Lensing Real Test: LR vs ΛCDM\n")

# Parameters
Om0 = 0.315
z = np.linspace(0.0, 4.0, 1000)          # Higher range for better accuracy
a = 1 / (1 + z)

# Hubble parameters
E_lcdm = np.sqrt(Om0 * (1+z)**3 + (1 - Om0))

# Logic Relativity (LR) - Customize here
exotic_factor = 1.0 + 0.08 * np.exp(-2.5 * a) * np.sin(4 * np.pi * np.log(a + 0.05))
E_lr = np.sqrt(Om0 * (1+z)**3 + (1 - Om0) * exotic_factor)

# Comoving distance χ(z)
def comoving_distance(z_vals, E_func):
    integrand = 1.0 / E_func
    chi = cumulative_trapezoid(integrand, z_vals, initial=0.0)
    return chi

chi_lcdm = comoving_distance(z, E_lcdm)
chi_lr   = comoving_distance(z, E_lr)

# ================================================
# Einstein Radius (SIS Lens)
# ================================================
def einstein_radius(z_l, z_s, chi_func):
    chi_l = np.interp(z_l, z, chi_func)
    chi_s = np.interp(z_s, z, chi_func)
    chi_ls = chi_s - chi_l

    D_l  = chi_l / (1 + z_l)
    D_s  = chi_s / (1 + z_s)
    D_ls = chi_ls / (1 + z_s)

    sigma_v = 200.0  # km/s
    theta_E = (4 * np.pi * (sigma_v / 3e5)**2 * (D_ls / D_s)) * 206265
    return max(theta_E, 0.01)   # Ensure positive value

# Test cases
print("=== STRONG LENSING EXAMPLES (Einstein Radius θ_E) ===")
for zl, zs in [(0.5, 2.0), (1.0, 3.0), (0.3, 1.5), (0.8, 2.5)]:
    print(f"z_l={zl}, z_s={zs} → ΛCDM: {einstein_radius(zl, zs, chi_lcdm):.2f} arcsec")
    print(f"z_l={zl}, z_s={zs} → LR   : {einstein_radius(zl, zs, chi_lr):.2f} arcsec\n")

# Plot
z_l_array = np.linspace(0.1, 2.0, 100)
theta_lcdm = [einstein_radius(zl, 2.0, chi_lcdm) for zl in z_l_array]
theta_lr   = [einstein_radius(zl, 2.0, chi_lr)   for zl in z_l_array]

plt.figure(figsize=(11, 7))
plt.plot(z_l_array, theta_lcdm, 'b-', linewidth=2.5, label='ΛCDM')
plt.plot(z_l_array, theta_lr,   'r--', linewidth=2.5, label='Logic Relativity (LR)')
plt.xlabel('Lens Redshift z_l')
plt.ylabel('Einstein Radius θ_E (arcsec)')
plt.title('Strong Gravitational Lensing Comparison\n(Source at z_s = 2.0)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("✅ Now working! Tune the 'exotic_factor' with your real LR physics.")
```
