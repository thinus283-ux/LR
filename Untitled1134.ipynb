```python
# ================================================
# LR Model - Deriving Π from Coupling (Stable Version)
# ================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ================== Parameters ==================
n_val = 1.1
kappa_val = 0.05      # <<< Smaller value avoids issues (tune 0.01 - 0.1)
beta_val = 2.0        # This controls γ ≈ n * beta
X_bg = 0.1            # Background kinetic term (positive)

rho_ratio = np.logspace(0, 4, 600)

# Always positive arguments
arg_bg = X_bg
arg_coupled = X_bg + kappa_val * rho_ratio**beta_val

# Pressure terms (positive base)
p_bg = (arg_bg)**n_val
p_coupled = (arg_coupled)**n_val

Pi_num = p_coupled - p_bg

# ================== Power-law Fit ==================
def power_law(r, P0, g):
    return P0 * r**g

mask = (rho_ratio > 5) & (Pi_num > 0)

if np.sum(mask) > 20:
    popt, _ = curve_fit(power_law, rho_ratio[mask], Pi_num[mask], p0=[1e-3, 2.2], bounds=(0, [np.inf, 5]))
    fitted_gamma = popt[1]
else:
    fitted_gamma = np.nan

print("=== DERIVATION RESULTS ===")
print(f"n          = {n_val}")
print(f"beta       = {beta_val}")
print(f"kappa      = {kappa_val}")
print(f"Fitted γ   ≈ {fitted_gamma:.3f}")
print(f"Π(ρ/ρ_bar=10) ≈ {np.interp(10, rho_ratio, Pi_num):.2e}")
print(f"Π(ρ/ρ_bar=100)≈ {np.interp(100, rho_ratio, Pi_num):.2e}")

# ================== Plot ==================
plt.figure(figsize=(10, 7))
plt.loglog(rho_ratio, Pi_num, 'b-', linewidth=2.5, label='Derived Π(ρ_b) from coupling')
if not np.isnan(fitted_gamma):
    plt.loglog(rho_ratio[mask], power_law(rho_ratio[mask], *popt), 'r--',
               label=f'Fit: γ = {fitted_gamma:.2f}')
plt.xlabel('ρ_b / ρ_bar')
plt.ylabel('Extra pressure Π')
plt.title('Π Derived from K-essence + Baryon Coupling')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
