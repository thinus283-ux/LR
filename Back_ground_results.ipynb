```python
import numpy as np
import matplotlib.pyplot as plt

Om0 = 0.315
w0  = -1.0
z_claim = 0.67

a = np.linspace(0.05, 2.0, 600)
z = 1/a - 1

E2 = Om0 * a**(-3) + (1 - Om0)
E = np.sqrt(E2)

Om_a = Om0 * a**(-3) / E2
Ode_a = (1 - Om0) / E2

q = 0.5 * Om_a - Ode_a

# Results at a=1 (z=0)
idx0 = np.argmin(np.abs(a - 1.0))
print("=== FINAL BACKGROUND RESULTS (at z=0) ===")
print(f"Ω_m,0              = {Om0}")
print(f"w0                 = {w0}")
print(f"Acceleration onset at z ≈ {z[np.argmin(np.abs(q))]:.3f}")
print(f"q(z=0)             = {q[idx0]:.4f}")
print(f"H(z=0)/H0          = {E[idx0]:.4f}  ✓")

# Plots omitted for brevity — same as before
```
