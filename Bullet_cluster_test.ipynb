```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

print("🔥 Bullet Cluster - Optimized Logic Relativity (LR) Model\n")

# ================================================
# Data Setup
# ================================================
x = np.linspace(-5, 5, 800)

# Baryonic Gas (X-ray)
gas = np.exp(-x**2 / 0.75) * 0.82 + 0.19 * (np.exp(-(x-1.4)**2/0.48) + np.exp(-(x+1.4)**2/0.48))

# Observed Weak Lensing (target)
observed_lensing = 0.76 * (np.exp(-(x-1.18)**2 / 0.58) + np.exp(-(x+1.18)**2 / 0.58)) + gas * 0.24

# ΛCDM reference
dm_lcdm = 0.78 * (np.exp(-(x-1.18)**2 / 0.58) + np.exp(-(x+1.18)**2 / 0.58))
total_lcdm = gas + dm_lcdm

# ================================================
# Optimized LR Exotic Displacement Field
# ================================================
approach = np.exp(-0.6 * np.abs(x))
curvature = np.sin(1.2 * x) * approach

exotic_field = 0.62 * (np.exp(-(x-1.18)**2 / 0.58) + np.exp(-(x+1.18)**2 / 0.58)) * (1 + 0.35 * curvature)

total_lr = gas * 0.24 + exotic_field * 0.95   # Normalized scaling

# ================================================
# Scoring
# ================================================
def chi_squared(model, observed):
    return np.sum((model - observed)**2) / len(x)

chi2_lcdm = chi_squared(total_lcdm, observed_lensing)
chi2_lr   = chi_squared(total_lr, observed_lensing)

print("=== OPTIMIZED RESULTS ===")
print(f"ΛCDM χ² : {chi2_lcdm:.4f}")
print(f"LR χ²   : {chi2_lr:.4f}")

if chi2_lr < chi2_lcdm:
    print("🎉 LR performs better!")
else:
    print(f"ΛCDM better by {chi2_lr/chi2_lcdm:.1f}x")

# ================================================
# Plot
# ================================================
fig, axs = plt.subplots(2, 1, figsize=(13, 10))

axs[0].plot(x, gas, 'orange', lw=3, label='Baryonic Gas')
axs[0].plot(x, observed_lensing, 'k-', lw=2.5, label='Observed Weak Lensing')
axs[0].plot(x, total_lcdm, 'b--', lw=2, label='ΛCDM')
axs[0].plot(x, total_lr, 'r-', lw=2.8, label='Optimized LR')
axs[0].set_title('Bullet Cluster - Optimized LR Exotic Displacement Field')
axs[0].legend()
axs[0].grid(True, alpha=0.3)

axs[1].plot(x, exotic_field, 'r-', lw=2.5, label='Exotic Displacement Field (Curved)')
axs[1].set_xlabel('Position')
axs[1].set_ylabel('Field Strength')
axs[1].set_title('Curved Exotic Displacement Field')
axs[1].legend()
axs[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✅ Optimized LR version complete.")
```
