```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# ====================== IMPROVED CONFIG ======================
N = 1500
r_max = 3.0
dr = r_max / (N - 1)
dt = 0.32 * dr
steps = 25000
t_max = steps * dt

r = np.linspace(0, r_max, N)

# Layered density (Diamond Core → Condensate → Thinning Shell)
rho = np.where(r < 0.55, 1.0,
        np.where(r < 1.52, 0.45,
                 0.04 / (r**1.1 + 0.2)))

c_eff = np.clip(1.0 / np.sqrt(rho + 0.07), 0.3, 2.8)

u = np.zeros(N)
v = np.zeros(N)

# Strong initial pulse
pulse_center = 95
sigma = 0.016
idx = slice(pulse_center, pulse_center + 180)
u[idx] = np.exp(-((r[idx] - r[pulse_center + 40])**2) / (2 * sigma**2))

# ====================== SIMULATION ======================
for step in range(steps):
    lap = (u[2:] - 2*u[1:-1] + u[:-2]) / dr**2
    v[1:-1] += dt * (c_eff[1:-1]**2 * lap)
    u[1:-1] += dt * v[1:-1]

    u[0] = u[1]                    # Hard reflection at Diamond Core
    u[-1] *= 0.965                 # Outer shell absorption

    # Condensate Haze damping (reduced for clearer echoes)
    shell = (r > 1.15) & (r < 1.65)
    u[shell] *= 0.996

# ====================== WOBBLE & OUTPUT ======================
times = np.linspace(0, t_max, steps)
wobble = 1 + 0.15 * np.sin(1.8 * times)

strain = u[750] * wobble
strain /= np.max(np.abs(strain)) + 1e-8   # normalize

# ====================== PLOTS ======================
fig, axs = plt.subplots(2, 1, figsize=(14, 9))

axs[0].plot(times, strain, 'b-', linewidth=1.4, label='GW Strain (Echo Train + Wobble)')
axs[0].set_title('Improved FDTD Echo-Cavity Simulation — Exotic Displacement Field Theory')
axs[0].set_xlabel('Time (ms)')
axs[0].set_ylabel('Normalized Strain h(t)')
axs[0].grid(True, alpha=0.3)
axs[0].legend()

freq = fftfreq(steps, dt)[:steps//2]
spectrum = np.abs(fft(strain))[:steps//2]
axs[1].plot(freq, spectrum, 'r-', linewidth=1.2)
axs[1].set_title('Frequency Spectrum (Wobble Sidebands Visible)')
axs[1].set_xlabel('Frequency (1/ms)')
axs[1].set_ylabel('Amplitude')
axs[1].set_xlim(0, 10)
axs[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('echo_fdtd_improved.png', dpi=400, bbox_inches='tight')
plt.show()

print("✅ Improved simulation complete! Clearer echo train generated.")
```
