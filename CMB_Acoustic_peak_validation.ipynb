import camb
import numpy as np
import matplotlib.pyplot as plt

def get_cmb_spectrum(H0_val):
    """
    Computes CMB power spectrum using CAMB.
    H0_val: Hubble constant (simulating modified expansion in LR)
    """
    pars = camb.CAMBparams()
    # Standard Planck 2018 parameters
    pars.set_cosmology(H0=H0_val, ombh2=0.02237, omch2=0.12, mnu=0.06, omk=0, tau=0.0544)
    pars.InitPower.set_params(ns=0.9649, As=2e-9)
    pars.set_for_lmax(2500)
    
    results = camb.get_results(pars)
    powers = results.get_cmb_power_spectra(pars, CMB_unit='muK')
    return powers['total']

# Execute simulations
ells = np.arange(2, 2501)
standard_model = get_cmb_spectrum(H0_val=67.4)
lr_modified_model = get_cmb_spectrum(H0_val=72.0) # Example of LR effect

# Visualization
plt.figure(figsize=(10, 6))
plt.plot(ells, standard_model[:, 0], color='black', alpha=0.5, label='Standard LCDM')
plt.plot(ells, lr_modified_model[:, 0], color='red', linestyle='--', label='LR Modified Expansion')

plt.title("Cosmic Microwave Background: Acoustic Peak Validation")
plt.xlabel("Multipole moment (l)")
plt.ylabel(r"Power $D_{\ell}$ [$\mu K^2$]")
plt.xlim(0, 2500)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
