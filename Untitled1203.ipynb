```python

# ================================================
# LR v2 - Morphology-Adaptive + Under 5 km/s Count
# ================================================

import numpy as np
import zipfile
import os
from scipy.optimize import curve_fit
import warnings
warnings.filterwarnings("ignore")

def stress_profile(r, P0, beta, r_max, size_factor):
    base = P0 * np.exp(-beta * r)
    rim_power = 2.85 + 0.35 * size_factor
    rim = 1 + 9.5 * (r / r_max)**rim_power
    return base * rim

def vortex_clamping(r, sigma, sigma_c, r_max, K, n, trans_fraction):
    excess = np.maximum(sigma - sigma_c, 0)
    r_trans = r_max * trans_fraction
    r_scale = np.maximum(r_max / 4.5, 1.2)
    rim_boost = 1 + 5.2 / (1 + np.exp(-(r - r_trans) / r_scale))
    return K * (excess ** n) * rim_boost

def fit_sparc_galaxy(r, Vobs, Verr, Vgas, Vdisk, Vbul):
    r_max = np.max(r)
    size_factor = np.clip((r_max - 5) / 15, 0, 1.1)

    has_bulge = np.max(Vbul) > 0 if len(Vbul) > 0 else False
    max_disk = np.max(Vdisk) if len(Vdisk) > 0 else 0
    max_gas = np.max(Vgas) if len(Vgas) > 0 else 0

    def model(r, P0, beta, sigma_c, Yd, Yb, K, n, trans_fraction):
        Vbar = np.sqrt(np.maximum(Vgas**2 + Yd * Vdisk**2 + Yb * Vbul**2, 0))
        sigma = stress_profile(r, P0, beta, r_max, size_factor)
        v_vortex = vortex_clamping(r, sigma, sigma_c, r_max, K, n, trans_fraction)
        return np.sqrt(Vbar**2 + v_vortex**2)

    if has_bulge and np.max(Vbul) > 0.5 * max_disk:
        p0 = [45000, 0.75, 1e-11, 0.7, 0.8, 5.2, 0.42, 0.40]
        trans_min, trans_max = 0.20, 0.65
    elif max_disk > 2.0 * max_gas:
        p0 = [28000, 0.45, 4e-12, 0.55, 0.0, 4.8, 0.38, 0.55]
        trans_min, trans_max = 0.35, 0.75
    else:
        p0 = [12000, 0.25, 8e-13, 0.2, 0.0, 3.5, 0.32, 0.70]
        trans_min, trans_max = 0.50, 0.90

    bounds = (
        [100, 0.005, 1e-16, 0.0, 0.0, 0.5, 0.05, trans_min],
        [350000, 6.0, 1e-4, 3.0, 3.0, 20.0, 0.95, trans_max]
    )

    try:
        popt, _ = curve_fit(
            model, r, Vobs,
            p0=p0,
            sigma=Verr,
            bounds=bounds,
            absolute_sigma=True,
            loss='soft_l1',
            maxfev=80000
        )
        V_fit = model(r, *popt)
        rms = np.sqrt(np.mean((V_fit - Vobs)**2))
        return rms, popt
    except:
        return np.nan, None

# ====================== FULL RUN WITH STATS ======================
print("Downloading official Rotmod_LTG.zip ...")
!wget -q https://astroweb.case.edu/SPARC/Rotmod_LTG.zip -O Rotmod_LTG.zip

with zipfile.ZipFile("Rotmod_LTG.zip", 'r') as zip_ref:
    zip_ref.extractall("sparc_data")

dat_files = [f for f in os.listdir("sparc_data") if f.endswith('.dat')]
print(f"Processing {len(dat_files)} real SPARC galaxies with LR v2...\n")

rms_list = []
under_5_count = 0
under_3_count = 0
under_10_count = 0

for fname in dat_files:
    path = os.path.join("sparc_data", fname)
    try:
        data = np.loadtxt(path, skiprows=1)
        if data.shape[1] < 6: continue

        r     = data[:, 0]
        Vobs  = data[:, 1]
        Verr  = data[:, 2]
        Vgas  = data[:, 3]
        Vdisk = data[:, 4]
        Vbul  = data[:, 5]

        rms, popt = fit_sparc_galaxy(r, Vobs, Verr, Vgas, Vdisk, Vbul)
        if not np.isnan(rms):
            rms_list.append(rms)
            if rms < 5:
                under_5_count += 1
            if rms < 3:
                under_3_count += 1
            if rms < 10:
                under_10_count += 1
    except:
        continue

rms_list = np.array(rms_list)

print("\n=== LR v2 Final Results ===")
print(f"Galaxies successfully fitted: {len(rms_list)}")
print(f"Median RMS: {np.nanmedian(rms_list):.2f} km/s")
print(f"Mean RMS:   {np.nanmean(rms_list):.2f} km/s")
print(f"Best: {np.nanmin(rms_list):.2f} km/s | Worst: {np.nanmax(rms_list):.2f} km/s")

print("\n=== How many galaxies under X km/s ===")
print(f"Under 3 km/s : {under_3_count}")
print(f"Under 5 km/s : {under_5_count}")
print(f"Under 10 km/s: {under_10_count}")
```
