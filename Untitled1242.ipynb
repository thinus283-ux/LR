```python

%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

print("Logic Relativity CMB Acoustic Validation")

PLANCK_ns = 0.9649
PLANCK_alpha_s = -0.0045
k_pivot = 0.05

def acoustic_power_spectrum(k, n_s, alpha_s):
    A_s = 2.1e-9
    return A_s * (k / k_pivot)**(n_s - 1 + 0.5 * alpha_s * np.log(k / k_pivot))

def compute_ns_array(k_values):
    tilt = 0.035 * (k_values / k_pivot)**(-0.1)
    return 1.0 - tilt

def compute_running(k_values, ns_array):
    log_k = np.log(k_values)
    alpha = np.gradient(ns_array, log_k)
    return np.mean(alpha)

k = np.logspace(-4, 1, 500)

ns_array = compute_ns_array(k)
ns_pred = np.mean(ns_array)
alpha_s_pred = compute_running(k, ns_array)
r_pred = 0.0

P_model = acoustic_power_spectrum(k, ns_pred, alpha_s_pred)

print(f"Model predictions:")
print(f"  n_s     = {ns_pred:.4f}   (Planck: {PLANCK_ns})")
print(f"  alpha_s = {alpha_s_pred:.5f}   (Planck: {PLANCK_alpha_s})")
print(f"  r       = {r_pred:.4f}")

def power_law(k, A, n):
    return A * (k / k_pivot)**(n - 1)

popt, _ = curve_fit(power_law, k, P_model, p0=[2.1e-9, 0.96])
print(f"Fitted n_s from P(k): {popt[1]:.4f}")

# Create figure with desired size
plt.figure(figsize=(10, 6))
plt.loglog(k, P_model, 'b-', lw=2, label='Logic Relativity Model')
plt.loglog(k, 2.1e-9 * (k/k_pivot)**(PLANCK_ns-1), 'r--', label='Planck 2018')
plt.xlabel(r'\( k \) [Mpc\( ^{-1} \)]')
plt.ylabel(r'\( P(k) \)')
plt.title('Primordial Power Spectrum Comparison')
plt.legend()
plt.grid(True, which='both', alpha=0.3)
plt.show()

plt.savefig('CMB_power_spectrum_comparison.png', dpi=300)
print("CMB validation complete. Plot saved as CMB_power_spectrum_comparison.png")
```
