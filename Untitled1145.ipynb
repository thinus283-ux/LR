```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid

# ================== FINAL PARAMETERS ==================
P0 = 1.2e-10
a_extra_strength = 0.012
additive_floor = 0.12e-10
r = np.logspace(-0.5, 2.7, 1400)

# Baryon profile
rho_b = 130 * np.exp(-r / 3.6) + 2200 * np.exp(-(r/0.65)**2)

# Accelerations
dr = np.gradient(r)
dm = 4 * np.pi * r**2 * rho_b * dr
M_enc = cumulative_trapezoid(dm, r, initial=0)
a_bary = (M_enc / (r**2 + 1e-8)) * (1.8e-10 / np.max(M_enc / (r**2 + 1e-8)))

# FIELD DYNAMICS: Sharp Handoff
# Use a steep sigmoid (1 / (1 + exp(-k*(r-r0)))) for a controlled transition
r0 = 35.0  # Transition radius
k = 0.5    # Sharpness
transition = 1.0 / (1.0 + np.exp(-k * (r - r0)))

dPi_dr = np.gradient(P0 * np.maximum(rho_b, 1.0)**2.2, r)
# Force is now a seamless sum: Baryons do the inner work, Dark Field does the outer work
a_extra = (a_extra_strength * (-dPi_dr / (np.max(rho_b)*0.05))) * (1 - transition) + (additive_floor * transition)
a_total = a_bary + a_extra

# Velocity
v_total = np.sqrt(np.abs(a_total) * r * 3.086e19) / 1000

# ================== FINAL OUTPUT ==================
print(f"{'Radius (kpc)':<15} | {'Velocity (km/s)':<15}")
print("-" * 35)
for p in [10.7, 29.7, 82.4, 228.6, 500.0]:
    v_val = np.interp(p, r, v_total)
    print(f"{p:<15.1f} | {v_val:<15.1f}")

plt.figure(figsize=(10, 6))
plt.semilogx(r, v_total, 'k-', linewidth=3, label='Calibrated Model')
plt.axhline(220, color='r', linestyle='--', label='Target (220 km/s)')
plt.xlabel('Radius [kpc]'); plt.ylabel('Velocity [km/s]')
plt.title('Final Stable Plateu: Controlled Handoff')
plt.grid(True, which="both", alpha=0.3); plt.legend(); plt.show()
```
