```python

import numpy as np
import matplotlib.pyplot as plt

print("🌌 Weak Lensing in Galaxy Clusters: Improved LR Model\n")

# Radius in Mpc
r = np.linspace(0.05, 3.5, 400)

# Baryonic component
baryon = 0.18 * np.exp(-r / 0.35) + 0.09 / (1 + (r / 0.9)**2)

# Observed / Target Weak Lensing profile (approximate real clusters)
observed = 1.85 * np.exp(-r / 0.45) + 0.35 / (1 + (r / 1.1)**2)

# ΛCDM reference (NFW-like DM halo)
dm_lcdm = 1.65 * np.exp(-r / 0.5)
total_lcdm = baryon + dm_lcdm

# ================================================
# Improved Logic Relativity (LR) Model
# ================================================
# Stronger Exotic Displacement Field for better central peak
exotic_amplitude = 1.35
exotic_field = exotic_amplitude * np.exp(-r / 0.48) * (1 + 0.45 * np.sin(3.2 * r) * np.exp(-1.8 * r))

total_lr = baryon * 1.22 + exotic_field * 1.15

print("=== WEAK LENSING COMPARISON (Improved) ===")
print(f"Peak convergence (center) - ΛCDM: {total_lcdm[0]:.3f}")
print(f"Peak convergence (center) - LR   : {total_lr[0]:.3f}")

# ================================================
# Plot
# ================================================
plt.figure(figsize=(12, 7))
plt.plot(r, baryon, 'orange', lw=2, label='Baryonic Matter')
plt.plot(r, observed, 'k-', lw=2.5, label='Observed Weak Lensing (target)')
plt.plot(r, total_lcdm, 'b-', lw=2.5, label='ΛCDM (Baryons + DM Halo)')
plt.plot(r, total_lr, 'r--', lw=3, label='Improved LR (Baryons + Exotic Displacement Field)')
plt.xlabel('Radius from Cluster Center (Mpc)')
plt.ylabel('Convergence / Projected Mass Density')
plt.title('Weak Lensing in Galaxy Clusters - Improved LR Model')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("\n✅ Improved Weak Lensing test complete.")
print("The Exotic Displacement Field has been tuned for better central peak matching.")
```
