```python
# ================================================
# LR Model - K-essence + Π Stability & Effective Parameters
# Corrected version
# ================================================

import numpy as np
import matplotlib.pyplot as plt

# ================== Parameters ==================
n = 1.1
alpha = 1.0      # tuning parameter
V0 = 1.0         # constant potential (normalized units)

# Positive kinetic variable K = -X  (K > 0)
K = np.logspace(-3, 1, 400)   # from very small to large kinetic term

# K-essence energy density and pressure
rho_phi = V0 + alpha * K**n * (2*n - 1)
p_phi   = -V0 + alpha * K**n

w_phi = p_phi / rho_phi

# Sound speed (standard K-essence result)
c_s2 = 1.0 / (2 * n - 1)
print(f"Base sound speed c_s² = {c_s2:.4f} ({c_s2*100:.1f}% of light speed)")

# ================== With Π compression ==================
P0 = 1.2e-10
gamma = 2.2
rho_ratio = np.logspace(0, 4, 200)   # from mean density to 10,000× overdensity

def c_s_eff(ratio):
    # Π makes the fluid stiffer (lower c_s) in overdensities
    Pi_term = P0 * ratio**gamma
    # Simple effective reduction (placeholder - tune with your derivation)
    return c_s2 / (1 + 5 * Pi_term)   # stronger suppression possible

c_s_eff_values = np.array([c_s_eff(r) for r in rho_ratio])

# ================== Results ==================
print("\n=== Stability Summary ===")
print("No ghosts          → Kinetic term positive: YES (α>0, n>0.5)")
print("Gradient stability → c_s² > 0: YES")
print("Causality          → c_s² < 1: YES (0.833 < 1)")
print(f"w_φ (relaxed limit, small K) ≈ {w_phi[0]:.4f}")
print(f"w_φ (high kinetic)           ≈ {w_phi[-1]:.4f}")

print("\nDensity dependence:")
for ratio in [1, 10, 100, 1000, 10000]:
    print(f"  ρ/ρ_bar = {ratio:5.0f}  →  c_s_eff² ≈ {c_s_eff(ratio):.4f}")

# ================== Plots ==================
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

axs[0].semilogx(K, w_phi, 'b-', linewidth=2)
axs[0].axhline(-1, color='k', ls='--', label='w=-1')
axs[0].set(xlabel='Kinetic term K = -X', ylabel='w_φ')
axs[0].set_title('Equation of State')
axs[0].grid(True, alpha=0.3)
axs[0].legend()

axs[1].semilogx(rho_ratio, c_s_eff_values, 'r-', linewidth=2)
axs[1].axhline(c_s2, color='b', ls='--', label=f'Pure K-essence c_s² = {c_s2:.3f}')
axs[1].set(xlabel='ρ_b / ρ_bar', ylabel='Effective c_s²')
axs[1].set_title('Density-dependent Sound Speed')
axs[1].grid(True, alpha=0.3)
axs[1].legend()

plt.tight_layout()
plt.show()
```
