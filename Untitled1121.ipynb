```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid

# ================== TUNING ==================
P0 = 1.2e-10
gamma = 2.2
a_extra_strength = 0.32      # <<< Change this (0.2 - 0.5)

r = np.logspace(-0.5, 2.6, 1200)   # 0.3 - 400 kpc

# Baryon density - pure exponential (simplest realistic case)
R_d = 3.5
rho_b = 120.0 * np.exp(-r / R_d)

# Π compression
Pi = P0 * np.maximum(rho_b, 1.0)**gamma

# Accelerations
dr = np.gradient(r)
dm = 4 * np.pi * r**2 * rho_b * dr
M_enc = cumulative_trapezoid(dm, r, initial=0)

a_bary_raw = M_enc / (r**2 + 1e-8)
a_bary = a_bary_raw * (2.5e-10 / np.max(a_bary_raw))

dPi_dr = np.gradient(Pi, r)
a_extra = a_extra_strength * (-dPi_dr / np.maximum(rho_b, 1e-12))

a_total = a_bary + a_extra

# Velocity conversion
def a_to_v(a, r_kpc):
    return np.sqrt(np.abs(a) * r_kpc * 3.086e19) / 1000

v_bary = a_to_v(a_bary, r)
v_extra = a_to_v(a_extra, r)
v_total = a_to_v(a_total, r)

# Results
print("=== FINAL ROTATION CURVE SUMMARY ===")
print(f"a_extra_strength = {a_extra_strength}")
print(f"a_extra @ 10 kpc   = {np.interp(10, r, a_extra):.2e} m/s²")
print(f"v_bary  @ 10 kpc   = {np.interp(10, r, v_bary):.1f} km/s")
print(f"v_extra @ 10 kpc   = {np.interp(10, r, v_extra):.1f} km/s")
print(f"v_total @ 10 kpc   = {np.interp(10, r, v_total):.1f} km/s")
print(f"v_total @ 30 kpc   = {np.interp(30, r, v_total):.1f} km/s")
print(f"v_total @ 100 kpc  = {np.interp(100, r, v_total):.1f} km/s")

# Full plots
fig, axs = plt.subplots(2, 2, figsize=(13, 9))

axs[0,0].loglog(r, rho_b, 'b-', label='ρ_b')
axs[0,0].set(xlabel='r [kpc]', ylabel='Density')
axs[0,0].legend(); axs[0,0].grid(True, alpha=0.3)

axs[0,1].semilogx(r, a_bary, 'b-', label='a_bary')
axs[0,1].semilogx(r, a_extra, 'r-', label='a_Π')
axs[0,1].semilogx(r, a_total, 'k-', label='a_total')
axs[0,1].set(xlabel='r [kpc]', ylabel='a (m/s²)')
axs[0,1].legend(); axs[0,1].grid(True, alpha=0.3)

axs[1,0].semilogx(r, v_bary, 'b-', label='Baryons only')
axs[1,0].semilogx(r, v_total, 'k-', linewidth=2.5, label='Baryons + Π')
axs[1,0].set(xlabel='r [kpc]', ylabel='v_c (km/s)')
axs[1,0].legend(); axs[1,0].grid(True, alpha=0.3)

axs[1,1].semilogx(r, v_extra, 'r-', label='v from Π')
axs[1,1].set(xlabel='r [kpc]', ylabel='v_extra (km/s)')
axs[1,1].legend(); axs[1,1].grid(True, alpha=0.3)

plt.suptitle('LR Model — Rotation Curve Test (Pure Exponential)')
plt.tight_layout()
plt.show()
```
