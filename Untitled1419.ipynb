```python

# =====================================================================
# LR-DC COSMOLOGY TEST — COMPLETE GOOGLE COLAB CELL
#
# Fixes:
#   1. Fortran source lines longer than 132 characters.
#   2. AxiECAMB's missing LambdaGeneral module build dependency.
#   3. Parallel Fortran module-build race by compiling sequentially.
#
# Runs:
#   A. Standard CDM / Lambda-CDM control
#   B. Quadratic scalar-field model: 1.0e-23 eV
#   C. Quadratic scalar-field model: 2.4e-22 eV
#   D. Quadratic scalar-field model: 1.0e-21 eV
#
# Downloads:
#   LRDC_AxiECAMB_results.zip
#
# IMPORTANT:
# These are exploratory linear cosmology comparisons.
# The scalar masses are trial values, not fitted LR-DC parameters.
# This does not itself derive KG mode occupations or close halo formation.
# =====================================================================

import os
import re
import time
import shutil
import zipfile
import subprocess
from pathlib import Path

# ----------------------- CONFIGURATION -----------------------

ROOT = Path("/content/LRDC_AxiECAMB_Colab")
REPO = ROOT / "AxiECAMB"
RESULTS = ROOT / "results"
ZIP_PATH = Path("/content/LRDC_AxiECAMB_results.zip")

H0 = 67.36
OMBH2 = 0.0224
OMDMH2 = 0.1200

REDSHIFTS = [249, 49, 0]

MODELS = [
    {
        "tag": "LRDC_LCDM_control",
        "kind": "lcdm",
        "mass_eV": 1.0e-27,
    },
    {
        "tag": "LRDC_scalar_m1e-23_eV",
        "kind": "scalar",
        "mass_eV": 1.0e-23,
    },
    {
        "tag": "LRDC_scalar_m2p4e-22_eV",
        "kind": "scalar",
        "mass_eV": 2.4e-22,
    },
    {
        "tag": "LRDC_scalar_m1e-21_eV",
        "kind": "scalar",
        "mass_eV": 1.0e-21,
    },
]


# ----------------------- HELPERS -----------------------

def run_checked(command, cwd=None, timeout=None):
    """Run a command, showing useful output if it fails."""
    print("\n$", " ".join(map(str, command)), flush=True)

    result = subprocess.run(
        command,
        cwd=cwd,
        text=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        timeout=timeout,
    )

    if result.stdout:
        print(result.stdout[-12000:], flush=True)

    if result.returncode != 0:
        raise RuntimeError(
            f"Command failed with exit code {result.returncode}: "
            + " ".join(map(str, command))
        )

    return result.stdout


def set_ini_value(text, key, value):
    """Replace an INI key or append it when it is absent."""
    pattern = re.compile(
        r"(?m)^\s*" + re.escape(key) + r"\s*=.*$"
    )
    replacement = f"{key} = {value}"

    if pattern.search(text):
        return pattern.sub(
            lambda match: replacement,
            text,
            count=1,
        )

    return text.rstrip() + "\n" + replacement + "\n"


def save_text(path, text):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(text)


# ----------------------- STEP 1 -----------------------
# Install build dependencies.

print("=" * 72)
print("STEP 1 — Installing compiler and libraries")
print("=" * 72)

run_checked([
    "bash", "-lc",
    "apt-get update -qq && "
    "apt-get install -y -qq "
    "git gfortran make liblapack-dev libblas-dev"
])


# ----------------------- STEP 2 -----------------------
# Start with a clean source checkout.

print("\n" + "=" * 72)
print("STEP 2 — Downloading AxiECAMB")
print("=" * 72)

if ROOT.exists():
    shutil.rmtree(ROOT)

if ZIP_PATH.exists():
    ZIP_PATH.unlink()

RESULTS.mkdir(parents=True, exist_ok=True)

run_checked([
    "git", "clone", "--depth", "1",
    "https://github.com/Ra-yne/AxiECAMB.git",
    str(REPO),
])

commit = subprocess.run(
    ["git", "rev-parse", "HEAD"],
    cwd=REPO,
    check=True,
    capture_output=True,
    text=True,
).stdout.strip()

save_text(
    ROOT / "source_commit.txt",
    f"AxiECAMB source commit:\n{commit}\n",
)

print("Source revision:", commit)


# ----------------------- STEP 3 -----------------------
# Fix long free-form Fortran source lines.

print("\n" + "=" * 72)
print("STEP 3 — Fixing Fortran line-length compiler option")
print("=" * 72)

makefile = REPO / "Makefile"

if not makefile.exists():
    raise FileNotFoundError("AxiECAMB Makefile was not found.")

make_text = makefile.read_text()

fflags_pattern = re.compile(r"(?m)^FFLAGS\s*=\s*(.*)$")
fflags_match = fflags_pattern.search(make_text)

if not fflags_match:
    raise RuntimeError(
        "Could not locate the active FFLAGS line in Makefile."
    )

flags = fflags_match.group(1).strip()

if "-ffree-line-length-none" not in flags:
    flags += " -ffree-line-length-none"

make_text = fflags_pattern.sub(
    "FFLAGS = " + flags,
    make_text,
    count=1,
)

makefile.write_text(make_text)

print("Compiler flags:", flags)

save_text(
    ROOT / "compiler_flags.txt",
    "FFLAGS = " + flags + "\n",
)


# ----------------------- STEP 4 -----------------------
# Fix missing dependency:
# halofit_ppf imports LambdaGeneral, defined in equations_ppf.f90.

print("\n" + "=" * 72)
print("STEP 4 — Fixing Fortran module dependency")
print("=" * 72)

main_makefile = REPO / "Makefile_main"

if not main_makefile.exists():
    raise FileNotFoundError("AxiECAMB Makefile_main was not found.")

main_text = main_makefile.read_text()

dependency_rule = (
    "$(NONLINEAR).o: $(EQUATIONS).o modules.o"
)

if dependency_rule not in main_text:
    main_text += (
        "\n\n"
        "# Colab build fix: the nonlinear module imports "
        "LambdaGeneral from the equations module.\n"
        + dependency_rule
        + "\n"
    )

main_makefile.write_text(main_text)

if dependency_rule not in main_makefile.read_text():
    raise RuntimeError(
        "Failed to verify the explicit Fortran module dependency."
    )

print("Added/verified dependency:")
print(dependency_rule)


# ----------------------- STEP 5 -----------------------
# Build sequentially: do not use make -j2.

print("\n" + "=" * 72)
print("STEP 5 — Cleaning and compiling sequentially")
print("=" * 72)

clean = subprocess.run(
    ["make", "clean"],
    cwd=REPO,
    text=True,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
)

print(clean.stdout[-4000:])

if clean.returncode != 0:
    raise RuntimeError("make clean failed.")

build_log = ROOT / "build.log"

print("Compiling with make -j1. Please wait...", flush=True)

with build_log.open("w") as log:
    build = subprocess.run(
        ["make", "-j1"],
        cwd=REPO,
        text=True,
        stdout=log,
        stderr=subprocess.STDOUT,
        timeout=1200,
    )

if build.returncode != 0:
    print("\nBUILD FAILED. Last part of build log:\n")
    print(build_log.read_text(errors="replace")[-18000:])

    raise RuntimeError(
        "AxiECAMB compilation failed. The full build log is stored at "
        + str(build_log)
    )

EXECUTABLE = REPO / "camb"

if not EXECUTABLE.exists():
    raise FileNotFoundError(
        "Compilation finished but the expected camb executable "
        "was not created."
    )

print("\nBUILD SUCCESSFUL — AxiECAMB executable found.")


# ----------------------- STEP 6 -----------------------
# Prepare model parameters.

print("\n" + "=" * 72)
print("STEP 6 — Preparing cosmology runs")
print("=" * 72)

base_params_file = REPO / "params.ini"

if not base_params_file.exists():
    raise FileNotFoundError("AxiECAMB params.ini was not found.")

base_params = base_params_file.read_text()

successful_models = []
failed_models = []


# ----------------------- STEP 7 -----------------------
# Run each cosmology with consistent background normalization.

for model in MODELS:

    tag = model["tag"]
    kind = model["kind"]
    mass = model["mass_eV"]

    print("\n" + "#" * 72)
    print("MODEL:", tag)
    print("#" * 72)

    print("Model type:", kind)
    print("Scalar mass [eV]:", f"{mass:.8e}")
    print("H0:", H0)
    print("ombh2:", OMBH2)
    print("Total DM abundance omdah2:", OMDMH2)

    params = base_params

    shared_settings = {
        "output_root": tag,

        "get_scalar_cls": "T",
        "get_vector_cls": "F",
        "get_tensor_cls": "F",
        "get_transfer": "T",

        "do_lensing": "T",
        "do_nonlinear": "0",

        "use_physical": "T",
        "hubble": f"{H0}",
        "ombh2": f"{OMBH2}",
        "omch2": f"{OMDMH2}",

        "m_ax": f"{mass:.8e}",

        "axion_isocurvature": "F",

        "transfer_num_redshifts": "3",
        "transfer_redshift(1)": str(REDSHIFTS[0]),
        "transfer_redshift(2)": str(REDSHIFTS[1]),
        "transfer_redshift(3)": str(REDSHIFTS[2]),

        "transfer_filename(1)": "transfer_z249.dat",
        "transfer_filename(2)": "transfer_z49.dat",
        "transfer_filename(3)": "transfer_z0.dat",

        "transfer_matterpower(1)": "matterpower_z249.dat",
        "transfer_matterpower(2)": "matterpower_z49.dat",
        "transfer_matterpower(3)": "matterpower_z0.dat",

        "transfer_kmax": "20",
        "transfer_k_per_logint": "8",
        "transfer_high_precision": "T",
        "transfer_interp_matterpower": "T",

        "accuracy_boost": "1",
        "l_accuracy_boost": "1",
        "movH_switch": "10",
        "feedback_level": "1",
    }

    for key, value in shared_settings.items():
        params = set_ini_value(params, key, value)

    if kind == "lcdm":

        # Standard CDM control: no scalar dark-matter abundance.
        params = set_ini_value(params, "use_axfrac", "F")
        params = set_ini_value(params, "omch2", f"{OMDMH2}")
        params = set_ini_value(params, "omaxh2", "0.0")

    elif kind == "scalar":

        # AxiECAMB interprets omdah2 as the total DM abundance
        # and axfrac as the scalar fraction when use_axfrac = T.
        # axfrac = 1 assigns the DM abundance to the scalar field.
        params = set_ini_value(params, "use_axfrac", "T")
        params = set_ini_value(params, "omdah2", f"{OMDMH2}")
        params = set_ini_value(params, "axfrac", "1.0")

    else:
        raise ValueError("Unknown model kind: " + str(kind))

    run_dir = RESULTS / tag
    run_dir.mkdir(parents=True, exist_ok=True)

    params_file = run_dir / "params.ini"
    params_file.write_text(params)

    log_file = run_dir / "solver.log"

    # Remove output files left behind by an earlier attempt with this tag.
    for old_file in REPO.glob(tag + "_*"):
        if old_file.is_file():
            old_file.unlink()

    start = time.time()

    try:
        print("Starting solver. This may take several minutes...", flush=True)

        with log_file.open("w") as log:
            run = subprocess.run(
                [str(EXECUTABLE), str(params_file)],
                cwd=REPO,
                stdout=log,
                stderr=subprocess.STDOUT,
                text=True,
                timeout=1200,
            )

        solver_log = log_file.read_text(errors="replace")

        if run.returncode != 0:
            raise RuntimeError(
                f"Solver returned exit code {run.returncode}"
            )

        # A normal solver return is not enough: confirm output exists.
        produced_files = [
            path for path in REPO.glob(tag + "_*")
            if path.is_file()
        ]

        if not produced_files:
            raise RuntimeError(
                "Solver returned without creating files prefixed "
                "with the model tag."
            )

        copied_files = []

        for source in produced_files:
            destination = run_dir / source.name
            shutil.copy2(source, destination)
            copied_files.append(destination)

        elapsed = (time.time() - start) / 60.0

        successful_models.append(tag)

        inventory = [
            f"{path.name}\t{path.stat().st_size} bytes"
            for path in sorted(copied_files)
        ]

        save_text(
            run_dir / "output_inventory.txt",
            "\n".join(inventory) + "\n",
        )

        print(f"Completed in {elapsed:.2f} minutes.")
        print("Output files collected:", len(copied_files))

        for path in sorted(copied_files):
            print("  ", path.name)

        # Record warnings, if any, for later review.
        warning_lines = [
            line for line in solver_log.splitlines()
            if "warning" in line.lower()
            or "error" in line.lower()
        ]

        save_text(
            run_dir / "warnings_and_errors.txt",
            (
                "\n".join(warning_lines) + "\n"
                if warning_lines
                else "No warning/error keyword lines detected.\n"
            ),
        )

    except Exception as exc:
        message = f"{type(exc).__name__}: {exc}"
        failed_models.append((tag, message))

        print("\nMODEL FAILED:", tag)
        print(message)

        if log_file.exists():
            print("\nLast part of solver log:")
            print(log_file.read_text(errors="replace")[-10000:])


# ----------------------- STEP 8 -----------------------
# Create a clear status report and explain scientific limitations.

print("\n" + "=" * 72)
print("STEP 8 — Writing report")
print("=" * 72)

report_lines = [
    "LR-DC / AxiECAMB Exploratory Cosmology Test",
    "=" * 48,
    "",
    f"AxiECAMB source revision: {commit}",
    "Compiler: GNU Fortran",
    "Fortran line-length option: -ffree-line-length-none",
    "Build mode: make -j1 (sequential)",
    "",
    "Shared normalization:",
    f"  H0 = {H0} km/s/Mpc",
    f"  ombh2 = {OMBH2}",
    f"  total DM abundance = {OMDMH2}",
    f"  requested redshifts = {REDSHIFTS}",
    "",
    "Models:",
]

for model in MODELS:
    report_lines.append(
        f"  {model['tag']}: "
        f"type={model['kind']}, "
        f"m_ax={model['mass_eV']:.8e} eV"
    )

report_lines += [
    "",
    "Results:",
]

for tag in successful_models:
    report_lines.append(f"  OUTPUTS FOUND: {tag}")

for tag, message in failed_models:
    report_lines.append(f"  FAILED: {tag} — {message}")

report_lines += [
    "",
    "Interpretation and limits:",
    "1. Trial scalar masses are exploratory values, not measured",
    "   or selected LR-DC parameters.",
    "",
    "2. AxiECAMB evolves a quadratic-potential ultralight scalar",
    "   model. It is not automatically equivalent to the LR-DC",
    "   complex Klein-Gordon branch or derived from Theory 2.5.",
    "",
    "3. These runs request matter transfer functions, matter-power",
    "   output, and scalar CMB spectra. Inspect file headers before",
    "   interpreting columns or using output in another solver.",
    "",
    "4. Do not assume standard matter-transfer files contain the",
    "   signed KG field perturbation, its time derivative, or the",
    "   velocity/current transfer required for a physical wave",
    "   initial condition. Do not erase signs or phases.",
    "",
    "5. This test does not establish a nonlinear halo, stable",
    "   rotating equilibrium, physical mode occupation, blind",
    "   SPARC predictions, or Planck/BAO/growth/lensing agreement.",
    "",
    "6. LR-DC Theory 2.5 remains unchanged by this comparison.",
    "",
]

if (
    len(successful_models) == len(MODELS)
    and not failed_models
):
    report_lines.append(
        "Wrapper status: all four runs produced prefixed output files."
    )
else:
    report_lines.append(
        "Wrapper status: INCOMPLETE — at least one model failed "
        "or produced no detected output."
    )

report = "\n".join(report_lines) + "\n"

save_text(RESULTS / "RUN_REPORT.txt", report)

print(report)


# ----------------------- STEP 9 -----------------------
# Package logs, parameter files, transfer outputs, and report.

print("\n" + "=" * 72)
print("STEP 9 — Creating results ZIP")
print("=" * 72)

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    compression=zipfile.ZIP_DEFLATED,
) as archive:
    for path in ROOT.rglob("*"):
        if path.is_file():
            archive.write(
                path,
                arcname=str(path.relative_to(ROOT.parent)),
            )

print("ZIP:", ZIP_PATH)
print("Size:", f"{ZIP_PATH.stat().st_size / 1e6:.2f} MB")


# ----------------------- STEP 10 -----------------------
# Final summary and download.

print("\n" + "=" * 72)
print("FINAL SUMMARY")
print("=" * 72)

print("Build: SUCCESS")
print("Runs producing output:", len(successful_models), "of", len(MODELS))

for tag in successful_models:
    print("OUTPUTS:", tag)

for tag, message in failed_models:
    print("FAILED:", tag, "—", message)

print("\nPlease retain the ZIP and every solver log.")

try:
    from google.colab import files
    files.download(str(ZIP_PATH))
except Exception as exc:
    print("Automatic download unavailable:", exc)
    print("Use Colab's Files panel to download:", ZIP_PATH)

print("\nFinished.")
```
