```python

# ================================================
# Logic Relativity - AIC & BIC vs ΛCDM (Clean & Reliable)
# ================================================

import numpy as np

print("✅ Logic Relativity AIC/BIC Test vs ΛCDM")
print("=" * 65)

# ==================  REPLACE THESE WITH YOUR REAL VALUES  ==================
# From your MCMC notebooks (e.g. Mcmc_Hubble_s8_tention.ipynb or SPARC fits)

k_LR   = 12          # Number of free parameters in your LR model
chi2_LR = 1450       # Best-fit chi-squared or -2*ln(L) for your model

k_LCDM = 6           # Standard ΛCDM parameters
chi2_LCDM = 1520     # Best-fit chi-squared for ΛCDM on same data

n_data = 8500        # Approximate total data points (SPARC + Planck + BAO etc.)
# =========================================================================

# Calculations
AIC_LR   = chi2_LR + 2 * k_LR
AIC_LCDM = chi2_LCDM + 2 * k_LCDM

BIC_LR   = chi2_LR + k_LR * np.log(n_data)
BIC_LCDM = chi2_LCDM + k_LCDM * np.log(n_data)

delta_AIC = AIC_LCDM - AIC_LR
delta_BIC = BIC_LCDM - BIC_LR

print(f"\n📊 RESULTS:")
print(f"AIC  (LR)   : {AIC_LR:.1f}")
print(f"AIC  (ΛCDM) : {AIC_LCDM:.1f}   → Delta AIC = {delta_AIC:.1f} {'✅ LR better' if delta_AIC > 0 else '❌'}")
print(f"BIC  (LR)   : {BIC_LR:.1f}")
print(f"BIC  (ΛCDM) : {BIC_LCDM:.1f}   → Delta BIC = {delta_BIC:.1f} {'✅ LR better' if delta_BIC > 0 else '❌'}")

if delta_AIC > 10:
    print("\n🎉 Strong evidence in favor of Logic Relativity (AIC)")
elif delta_AIC > 0:
    print("\n✅ LR preferred on AIC")
else:
    print("\n⚠️  ΛCDM preferred on AIC")

print("\nNote: Lower AIC/BIC = better model (penalizes extra parameters)")
```
